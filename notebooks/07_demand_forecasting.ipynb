# 📈 Nexora — Time Series Demand Forecasting
**Phase 8: Temporal Volume Aggregation, Seasonal Modeling & Inventory Planning**

### Objective
Forecast daily and weekly demand across product categories to optimize inventory stocking, prevent stockouts, and manage warehouse throughput:
- Construct continuous temporal timeline from customer order cadences.
- Benchmark simple baseline (**7-Day Moving Average**) against **Holt-Winters Exponential Smoothing** and **Seasonal ARIMA (SARIMA)**.
- Evaluate holdout backtest accuracy using **MAE**, **RMSE**, and **MAPE**.
- Produce 30-day forward demand projections with 95% confidence intervals.

```python
import json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid", palette="deep")
DATA_DIR = Path("../data/processed")

with open(DATA_DIR / "forecasting_evaluation_metrics.json", "r") as f:
    eval_data = json.load(f)

df_eval = pd.DataFrame(eval_data["forecasting_benchmark"])
print("=== Time Series Forecasting Benchmark Results (30-Day Holdout) ===")
display(df_eval)
```

---
## ❓ Business Question 1: How do seasonal time-series models compare against simple moving average baselines?
*Compare MAE, RMSE, and MAPE error metrics across models.*

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# MAE Comparison
sns.barplot(data=df_eval, x="model", y="mae", ax=axes[0], palette="Blues_r")
axes[0].set_title("Mean Absolute Error (MAE) Comparison (Lower is Better)", fontsize=13, fontweight="bold")
axes[0].set_ylabel("MAE (Orders / Day)")
axes[0].tick_params(axis='x', rotation=15)

# MAPE Comparison
sns.barplot(data=df_eval, x="model", y="mape", ax=axes[1], palette="Reds_r")
axes[1].set_title("Mean Absolute Percentage Error (MAPE %)", fontsize=13, fontweight="bold")
axes[1].set_ylabel("MAPE (%)")
axes[1].tick_params(axis='x', rotation=15)

plt.tight_layout()
plt.show()
```

---
## ❓ Business Question 2: What is the projected daily demand for the upcoming 30 days?
*Review the 30-day forward demand trajectory and 95% confidence intervals for warehouse operations.*

```python
df_forecast = pd.read_parquet(DATA_DIR / "demand_forecasts.parquet")
df_forecast["forecast_date"] = pd.to_datetime(df_forecast["forecast_date"])

plt.figure(figsize=(13, 5))
plt.plot(df_forecast["forecast_date"], df_forecast["predicted_demand"], marker='o', color='#1d3557', label="Predicted Daily Orders", linewidth=2)
plt.fill_between(df_forecast["forecast_date"], df_forecast["lower_ci_95"], df_forecast["upper_ci_95"], color='#457b9d', alpha=0.25, label="95% Confidence Interval")

plt.title("30-Day Operational Forward Demand Forecast", fontsize=14, fontweight="bold")
plt.xlabel("Forecast Date")
plt.ylabel("Projected Order Volume")
plt.xticks(rotation=30)
plt.legend()
plt.show()
```
